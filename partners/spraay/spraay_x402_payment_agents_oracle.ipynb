## Overview

AI agents can reason, retrieve context, and generate responses — but they can't **pay for resources** autonomously. As agents move toward production, they need the ability to pay for premium inference, external data, and services without human intervention.

This notebook demonstrates how to combine:
- **Oracle AI Database** — vector storage, semantic search, and agent memory
- **Spraay x402 Gateway** — autonomous USDC micropayments over HTTP (the [x402 protocol](https://www.x402.org))
- **LangChain** — agent orchestration and tool use

The result is an agent that retrieves context from Oracle's vector store, decides when it needs premium AI inference, and **pays for it automatically** via Spraay's x402 gateway — all in a single agentic loop.

### What is x402?

x402 is an open payment protocol built on HTTP status code 402 ("Payment Required"). When an agent requests a paid resource, the server responds with a `402` status and payment requirements. The agent's payment middleware automatically signs a USDC payment on Base and retries the request with a payment proof header. No API keys, no subscriptions — just pay-per-request.

### What is Spraay?

[Spraay Protocol](https://spraay.app) is a multi-chain batch payment protocol. The chains it is deployed on are listed in the [contract deployments](https://docs.spraay.app#contract-deployments) section of the docs. The [Spraay x402 Gateway](https://gateway.spraay.app) exposes pay-per-call endpoints for AI inference, Bittensor decentralized AI, search/RAG, batch payments, escrow, RPC, IPFS storage, robotics (RTP), supply chain (SCTP), agent wallets (ERC-4337), and more. The current endpoints, categories and prices are published in the gateway's live catalog: [`/.well-known/x402.json`](https://gateway.spraay.app/.well-known/x402.json).

Spraay's batch payment tool is an official community tool in [Google's Agent Development Kit (ADK)](https://github.com/google/adk-python-community) — merged as PR #95.

## 5. Build the x402 Payment Client

The x402 payment flow works like this:

1. Agent makes an HTTP request to a paid endpoint
2. Server responds with `402 Payment Required` and payment requirements (price, recipient, token) in `accepts[]`
3. Client signs an EIP-3009 `transferWithAuthorization` for USDC on Base
4. Client retries the request with the signed payment in the `PAYMENT-SIGNATURE` header
5. Server verifies payment and returns the response

Rather than hand-rolling the signing, we use the official [`x402`](https://pypi.org/project/x402/) package: it selects a requirement from `accepts[]`, signs the EIP-712 payload on the USDC-on-Base domain, and encodes the header.

> **Note on the HTTP client**: `x402`'s httpx integration is async-only, so the synchronous LangChain tools below use its `requests` session (`x402_requests`) instead.


````python
import base64
import json

from eth_account import Account
from x402 import x402ClientSync
from x402.http.clients import x402_requests
from x402.mechanisms.evm import EthAccountSigner
from x402.mechanisms.evm.exact.register import register_exact_evm_client

# USDC on Base uses 6 decimals, so an amount of "5000" atomic units is $0.005.
USDC_DECIMALS = 6


class SpraayX402Client:
    """Client for making x402 payments to the Spraay Gateway.

    Gateway: https://gateway.spraay.app
    Docs: https://docs.spraay.app
    Live endpoint catalog: https://gateway.spraay.app/.well-known/x402.json

    Payment is delegated to the official ``x402`` package: it reads the quote
    from ``accepts[]``, signs an EIP-3009 ``transferWithAuthorization`` on the
    USDC-on-Base EIP-712 domain, and attaches the base64 ``PAYMENT-SIGNATURE``
    header on the retry.

    Signing a payment is not the same as spending: the gateway settles only
    when its handler succeeds. A payment therefore counts toward the total
    only once the ``PAYMENT-RESPONSE`` receipt confirms settlement, and the
    receipt's transaction hash is kept so every charge is verifiable on Base.
    """

    def __init__(self, gateway_url: str, private_key: str | None):
        self.gateway_url = gateway_url.rstrip("/")
        self.payment_log = []
        self.account = None
        self.session = None
        self._pending_quote = None

        # Without a wallet key the free cells still run; paid calls return a
        # clear message instead of failing deep inside the signing library.
        if private_key:
            self.account = Account.from_key(private_key)
            self.client = x402ClientSync()
            self.client.on_after_payment_creation(self._record_quote)
            register_exact_evm_client(self.client, EthAccountSigner(self.account))
            self.session = x402_requests(self.client)

    def _record_quote(self, context) -> None:
        """Remember the quote that was signed, to pair with the settlement receipt."""
        selected = context.selected_requirements
        resource = getattr(context.payment_required, "resource", None)
        self._pending_quote = {
            "endpoint": getattr(resource, "url", None),
            "amount": selected.amount,  # atomic units, e.g. "5000"
            "asset": selected.asset,
            "network": selected.network,
            "pay_to": selected.pay_to,
        }

    def _record_settlement(self, response) -> None:
        """Record the outcome of a payment attempt, settled or not."""
        raw = (response.headers.get("PAYMENT-RESPONSE")
               or response.headers.get("X-PAYMENT-RESPONSE"))
        if self._pending_quote is None and raw is None:
            return  # no payment was attempted on this request

        receipt = None
        if raw:
            try:
                receipt = json.loads(base64.b64decode(raw))
            except Exception:
                receipt = {"raw": raw}

        entry = dict(self._pending_quote or {})
        entry["settled"] = bool(receipt and receipt.get("success"))
        entry["tx"] = (receipt or {}).get("transaction")
        entry["http_status"] = response.status_code
        self.payment_log.append(entry)

    def request(self, endpoint: str, method: str = "GET", **kwargs) -> dict:
        """Make an x402-enabled request to the Spraay Gateway."""
        if self.session is None:
            return {
                "error": {
                    "status": None,
                    "body": {
                        "message": (
                            "WALLET_PRIVATE_KEY is not set, so paid endpoints are "
                            "disabled. Add a funded EVM wallet private key "
                            "(0x-prefixed 32-byte hex) to your .env to enable "
                            "x402 payments."
                        )
                    },
                }
            }

        url = f"{self.gateway_url}{endpoint}"
        self._pending_quote = None
        response = self.session.request(method, url, timeout=30, **kwargs)
        self._record_settlement(response)

        if response.status_code == 200:
            return response.json()

        # Return the gateway's own error body verbatim, not a bare status string.
        try:
            body = response.json()
        except ValueError:
            body = {"raw": response.text}
        return {"error": {"status": response.status_code, "body": body}}

    def get_payment_summary(self) -> dict:
        """Get a summary of this session. Only settled payments count as spend."""
        total_atomic = 0
        settled = 0
        for payment in self.payment_log:
            if not payment.get("settled"):
                continue
            settled += 1
            amount = payment.get("amount")
            if amount is None:
                continue
            try:
                total_atomic += int(amount)
            except (TypeError, ValueError):
                continue
        return {
            "total_attempts": len(self.payment_log),
            "settled_payments": settled,
            "total_usdc_atomic": total_atomic,
            "total_usdc_spent": total_atomic / 10**USDC_DECIMALS,
            "payments": self.payment_log,
        }


# Initialize the x402 client
x402_client = SpraayX402Client(
    gateway_url=SPRAAY_GATEWAY_URL,
    private_key=WALLET_PRIVATE_KEY
)

if x402_client.account is None:
    print("⚠️  x402 client created without a wallet key — paid endpoints are disabled")
    print("   Set WALLET_PRIVATE_KEY in .env to enable x402 payments")
    print(f"   Gateway: {SPRAAY_GATEWAY_URL}")
else:
    print("✅ x402 Client initialized")
    print(f"   Wallet: {x402_client.account.address}")
    print(f"   Gateway: {SPRAAY_GATEWAY_URL}")
````

## 6. Define Agent Tools

Now we create LangChain tools that the agent can use:

1. **`search_knowledge_base`** — Retrieves relevant context from Oracle AI Database (free)
2. **`spraay_ai_inference`** — Calls AI models via Spraay's x402 gateway ($0.005 USDC/request)
3. **`spraay_web_search`** — Searches the web via Tavily through Spraay ($0.02 USDC/request)

The agent decides which tools to use based on the query. Free retrieval first, paid inference only when needed.


````python
from langchain_core.tools import tool

@tool
def search_knowledge_base(query: str) -> str:
    """Search the Oracle AI Database vector store for relevant context.
    Use this FIRST to check if the answer exists in our knowledge base.
    This is a free operation — no payment required."""

    results = vector_store.similarity_search(query, k=3)

    if not results:
        return "No relevant documents found in the knowledge base."

    context = "\n\n".join([
        f"[Source: {doc.metadata.get('source', 'unknown')}] {doc.page_content}"
        for doc in results
    ])
    return f"Found {len(results)} relevant documents:\n\n{context}"


@tool
def spraay_ai_inference(prompt: str) -> str:
    """Call a premium AI model via the Spraay x402 Gateway.
    Costs $0.005 USDC per request. Uses /api/v1/chat/completions.
    Use ONLY when the knowledge base doesn't have the answer."""

    print(f"  🤖 Calling AI inference via Spraay x402 ($0.005 USDC)...")
    response = x402_client.request(
        "/api/v1/chat/completions",
        method="POST",
        json={
            "model": "openai/gpt-5.5",
            "messages": [{"role": "user", "content": prompt}],
            "max_tokens": 500
        }
    )

    if "error" in response:
        return f"Inference failed: {json.dumps(response['error'])}"

    return response.get("choices", [{}])[0].get("message", {}).get("content", "No response")


@tool
def spraay_bittensor_inference(prompt: str) -> str:
    """Call decentralized AI via Bittensor through Spraay's x402 Gateway.
    Costs $0.03 USDC per request. Uses /bittensor/v1/chat/completions.
    OpenAI-compatible drop-in routed through Bittensor SN64 (Chutes AI).
    Models include DeepSeek, Qwen and Mistral; the current list is available
    from /bittensor/v1/models."""

    print(f"  🧬 Calling Bittensor inference via Spraay x402 ($0.03 USDC)...")
    response = x402_client.request(
        "/bittensor/v1/chat/completions",
        method="POST",
        json={
            "model": "deepseek-ai/DeepSeek-V3.2-TEE",
            "messages": [{"role": "user", "content": prompt}],
            "max_tokens": 500
        }
    )

    if "error" in response:
        return f"Bittensor inference failed: {json.dumps(response['error'])}"

    return response.get("choices", [{}])[0].get("message", {}).get("content", "No response")


@tool
def spraay_web_search(query: str) -> str:
    """Search the web via Spraay's x402 search endpoint (Tavily-powered).
    Costs $0.02 USDC per request. Uses /api/v1/search/web.
    Use when you need current information not available in the knowledge base."""

    print(f"  🔍 Searching the web via Spraay x402 ($0.02 USDC)...")
    response = x402_client.request(
        "/api/v1/search/web",
        method="POST",
        json={"query": query, "search_depth": "basic", "max_results": 5}
    )

    if "error" in response:
        return f"Search failed: {json.dumps(response['error'])}"

    results = response.get("results", [])
    if not results:
        return "No web results found."

    formatted = "\n".join([
        f"- {r.get('title', 'Untitled')}: {r.get('content', 'No content')}"
        for r in results[:5]
    ])
    return f"Web search results:\n{formatted}"


tools = [search_knowledge_base, spraay_ai_inference, spraay_bittensor_inference, spraay_web_search]
print(f"✅ {len(tools)} agent tools registered")
print("   - search_knowledge_base      (free — Oracle AI DB)")
print("   - spraay_ai_inference         ($0.005 USDC — Spraay AI inference)")
print("   - spraay_bittensor_inference  ($0.03 USDC — Bittensor SN64)")
print("   - spraay_web_search           ($0.02 USDC — Tavily)")
````

## 7. Build the Payment-Enabled Agent

We create a LangChain ReAct agent that:
1. **Always checks the knowledge base first** (free)
2. **Escalates to paid inference** only when local context is insufficient
3. **Can choose between centralized AI or decentralized Bittensor inference**
4. **Logs all payments** for transparency and auditing

This cost-aware behavior is guided by the system prompt — the agent is instructed to prefer free tools and only pay when necessary.


````python
from langchain.agents import create_agent
from langchain.agents.middleware import ModelCallLimitMiddleware, ToolCallLimitMiddleware
from langchain_openai import ChatOpenAI

# Initialize LLM for agent reasoning
llm = ChatOpenAI(
    model="gpt-5.5",
    temperature=0,
    openai_api_key=OPENAI_API_KEY
)

# System prompt that enforces cost-aware behavior
system_prompt = """You are an AI research agent with access to both free and paid tools.

IMPORTANT RULES:
1. ALWAYS search the knowledge base first using search_knowledge_base (this is FREE)
2. Only use paid tools if the knowledge base doesn't have a sufficient answer
3. For AI analysis, prefer spraay_ai_inference ($0.005) for complex tasks or
   spraay_bittensor_inference ($0.03) for decentralized/censorship-resistant inference
4. Use spraay_web_search ($0.02) only for current/real-time information
5. When using paid tools, be specific in your queries to minimize cost
6. After answering, briefly note which tools you used and whether any payments were made"""

# The prompt tells the agent to use a paid tool only when it needs one, so each
# paid tool is capped at a single call per query. Without this the agent can
# re-query a paid endpoint many times for one question, and the cost-aware
# behaviour the notebook demonstrates would not hold. The free Oracle tool is
# deliberately left uncapped. ModelCallLimitMiddleware caps the reasoning loop
# the way max_iterations did in older LangChain.
paid_tool_caps = [
    ToolCallLimitMiddleware(tool_name=name, run_limit=1, exit_behavior="continue")
    for name in ("spraay_ai_inference", "spraay_bittensor_inference", "spraay_web_search")
]

agent_executor = create_agent(
    model=llm,
    tools=tools,
    system_prompt=system_prompt,
    middleware=[ModelCallLimitMiddleware(run_limit=5, exit_behavior="end"), *paid_tool_caps],
)

print("✅ Payment-enabled agent ready")
print("   Paid tools are capped at one call each per query")
````

## 9. Payment Audit Trail

One of the key benefits of x402 is full payment transparency. Every micropayment is an on-chain USDC transfer on Base, providing a complete audit trail. Let's review what the agent spent during this session.


````python
# Review all payments made during this session
summary = x402_client.get_payment_summary()


def format_amount(amount):
    """Amounts come from the gateway quote in atomic units, or are absent."""
    if amount is None:
        return "amount not recorded"
    try:
        return f"${int(amount) / 10**USDC_DECIMALS:.6f} USDC"
    except (TypeError, ValueError):
        return f"unrecognized amount: {amount!r}"


print("=" * 60)
print("💰 PAYMENT SUMMARY")
print("=" * 60)
print(f"Payment attempts:            {summary['total_attempts']}")
print(f"Settled on-chain:            {summary['settled_payments']}")
print(f"Total USDC spent:            ${summary['total_usdc_spent']:.6f}")
print()

if summary['payments']:
    print("Individual payments:")
    for i, payment in enumerate(summary['payments'], 1):
        endpoint = payment.get('endpoint') or 'unknown endpoint'
        print(f"  {i}. {endpoint} — {format_amount(payment.get('amount'))}")
        if payment.get('settled'):
            print(f"     settled: https://basescan.org/tx/{payment.get('tx')}")
        else:
            print(f"     NOT SETTLED — gateway returned HTTP {payment.get('http_status')};"
                  " the signed payment was never charged")
else:
    print("No paid requests were made (all answered from knowledge base)")

print()
if x402_client.account is not None:
    print("All settled payments are verifiable on-chain at:")
    print("https://basescan.org/address/" + x402_client.account.address)
else:
    print("No wallet configured — set WALLET_PRIVATE_KEY in .env to enable payments.")
````
